```python
most_recorded = (
    df
    .groupby(['devRef','name','extra'], as_index=False)
    .agg(total_watch_time=('duration', 'sum'))
    .sort_values(by='total_watch_time', ascending=False)
)

most_recorded
```

```python
import json

def extract(x):
    try:
        d = json.loads(x) if x else {}
        return (
            d.get('programId'),
            d.get('title'),
        )
    except:  # noqa: E722
        return (None, None)
```

```python
df_tmp = df[['devRef', 'name', 'duration', 'extra']].copy()

df_tmp[['program_id', 'title']] = pd.DataFrame(
    df_tmp['extra'].apply(extract).tolist(),
    index=df_tmp.index
)
```

```python
df_tmp = df_tmp.drop(columns=['extra'])
df_tmp = df_tmp.reset_index(drop=True)
df_tmp
```

```python
df_tmp = (
    df_tmp
    .groupby(['devRef', 'name', 'program_id', 'title' ], as_index=False)
    .agg(total_watch_time=('duration', 'sum'))
)

df_tmp
```

```python
df_tmp = (
    df_tmp
    .groupby(['program_id','name', 'title'])['devRef']
    .nunique()
    .reset_index(name='device_count')
    .sort_values(by='device_count', ascending=False)
)

df_tmp
```
